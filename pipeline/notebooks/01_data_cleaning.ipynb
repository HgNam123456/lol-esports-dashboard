```python
# Cell 5: Player stats aggregate từ player_df

# Kiểm tra cột
required_player_cols = [
    "gameid", "playername", "teamname", "result",
    "kills", "deaths", "assists",
    "dpm", "damageshare", "visionscore",
    "earnedgold", "earned gpm",  # trong file tên là "earned gpm" có khoảng trắng
]
missing_player = [c for c in required_player_cols if c not in player_df.columns]
print("Missing player cols:", missing_player)

# Đổi tên cột có khoảng trắng cho dễ xử lý
player_df = player_df.rename(columns={"earned gpm": "earned_gpm"})

agg_player = (
    player_df
    .groupby(["playername", "teamname"], as_index=False)
    .agg(
        games=("gameid", "nunique"),
        wins=("result", "sum"),
        kills=("kills", "sum"),
        deaths=("deaths", "sum"),
        assists=("assists", "sum"),
        dpm=("dpm", "mean"),
        damageshare=("damageshare", "mean"),
        visionscore=("visionscore", "mean"),
        earnedgold=("earnedgold", "mean"),
        earnedgpm=("earned_gpm", "mean"),
    )
)

agg_player["losses"] = agg_player["games"] - agg_player["wins"]
agg_player["winrate"] = agg_player["wins"] / agg_player["games"]

# KDA: tránh chia cho 0
agg_player["kda"] = (
    (agg_player["kills"] + agg_player["assists"])
    / agg_player["deaths"].replace(0, np.nan)
)
agg_player["kda"] = agg_player["kda"].fillna(
    agg_player["kills"] + agg_player["assists"]
)

players_df_final = agg_player
players_df_final.head()
```

```python
print(team_df.columns)

# 1) Bảng aggregate theo team cho KPI (teams_df_final)
agg_team = (
    team_df
    .groupby("teamname", as_index=False)
    .agg(
        games=("gameid", "nunique"),
        wins=("result", "sum"),
        gamelength=("gamelength", "mean"),
        dragons=("dragons", "mean"),
        opp_dragons=("opp_dragons", "mean"),
        barons=("barons", "mean"),
        opp_barons=("opp_barons", "mean"),
        heralds=("heralds", "mean"),
        opp_heralds=("opp_heralds", "mean"),
        towers=("towers", "mean"),
        opp_towers=("opp_towers", "mean"),
        inhibitors=("inhibitors", "mean"),
        opp_inhibitors=("opp_inhibitors", "mean"),
        turretplates=("turretplates", "mean"),
        opp_turretplates=("opp_turretplates", "mean"),
        firstdragon_rate=("firstdragon", "mean"),
        firstherald_rate=("firstherald", "mean"),
        firstbaron_rate=("firstbaron", "mean"),
        firsttower_rate=("firsttower", "mean"),
    )
)

agg_team["losses"] = agg_team["games"] - agg_team["wins"]
agg_team["winrate"] = agg_team["wins"] / agg_team["games"]

teams_df_final = agg_team

# 2) Bảng match-level cho history (team_matches_df)
#    Suy ra opponent từ gameid + teamname
base_cols = [
    "gameid",
    "year",
    "league",
    "teamname",
    "side",
    "result",
    "dragons",
    "barons",
    "heralds",
    "towers",
    "inhibitors",
]
base_cols = [c for c in base_cols if c in team_df.columns]

tmp = team_df[base_cols].copy()

# self-merge để lấy team còn lại trong cùng game
left = tmp.rename(columns={"teamname": "team", "result": "result_team"})
right = tmp.rename(
    columns={
        "teamname": "opponent",
        "result": "result_opponent",
    }
)

merged = left.merge(
    right[["gameid", "opponent", "result_opponent"]],
    on="gameid",
    how="left",
)

# loại trường hợp self-join (team == opponent)
merged = merged[merged["team"] != merged["opponent"]].copy()

team_matches_df = merged.rename(
    columns={
        "team": "teamname",
        "result_team": "result",
    }
)[
    [
        "gameid",
        "year",
        "league",
        "teamname",
        "side",
        "opponent",
        "result",
        "dragons",
        "barons",
        "heralds",
        "towers",
        "inhibitors",
    ]
]

teams_df_final.head(), team_matches_df.head()
```

```python
# Cell 7: Champion stats aggregate từ player_df

required_champ_cols = [
    "champion", "gameid", "result",
    "kills", "deaths", "assists",
    "dpm", "damageshare", "visionscore",
    "earnedgold", "earned_gpm",
    "firstblood",      # mới
    "golddiffat15",    # mới
]
missing_champ = [c for c in required_champ_cols if c not in player_df.columns]
print("Missing champion cols:", missing_champ)

champ_agg = (
    player_df
    .groupby("champion", as_index=False)
    .agg(
        games=("gameid", "nunique"),
        wins=("result", "sum"),
        kills=("kills", "mean"),
        deaths=("deaths", "mean"),
        assists=("assists", "mean"),
        dpm=("dpm", "mean"),
        damageshare=("damageshare", "mean"),
        visionscore=("visionscore", "mean"),
        goldpermin=("earned_gpm", "mean"),
        firstblood=("firstblood", "sum"),          # tổng số lần first blood
        golddiffat15=("golddiffat15", "mean"),     # gold diff @15 trung bình
    )
)

champ_agg["winrate"] = champ_agg["wins"] / champ_agg["games"]
champ_agg["firstblood_rate"] = champ_agg["firstblood"] / champ_agg["games"]

champions_df_final = champ_agg
champions_df_final.head()

```

```python
# team_df: chỉ chứa dòng team (position == "team")

total_matches = team_df["gameid"].nunique()
total_players = player_df["playername"].nunique()
total_teams = team_df["teamname"].nunique()
total_champions = player_df["champion"].nunique()

# Chuẩn hoá side về lowercase cho chắc
team_df["side_norm"] = team_df["side"].str.lower()

# Blue side
blue_teams = team_df[team_df["side_norm"] == "blue"]
blue_games = blue_teams["gameid"].nunique()
blue_wins = int((blue_teams["result"] == 1).sum())
blue_side_winrate = blue_wins / blue_games if blue_games > 0 else 0.0

# Red side
red_teams = team_df[team_df["side_norm"] == "red"]
red_games = red_teams["gameid"].nunique()
red_wins = int((red_teams["result"] == 1).sum())
red_side_winrate = red_wins / red_games if red_games > 0 else 0.0

print("blue_games:", blue_games, "blue_wins:", blue_wins)
print("red_games:", red_games, "red_wins:", red_wins)

overview_dict = {
    "total_matches": int(total_matches),
    "total_players": int(total_players),
    "total_teams": int(total_teams),
    "total_champions": int(total_champions),
    "blue_side_winrate": float(blue_side_winrate),
    "red_side_winrate": float(red_side_winrate),
}
overview_dict
```

```python
# Cell 9: Association rules cho player (demo đơn giản)
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Sample cho nhanh
sample = player_df.sample(min(len(player_df), 5000), random_state=42)

transactions = []
for _, row in sample.iterrows():
    items = []
    items.append(f"champ_{row['champion']}")
    items.append(f"role_{row['position']}")
    if row["result"] == 1:
        items.append("result_win")
    else:
        items.append("result_loss")
    transactions.append(items)

te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
trans_df = pd.DataFrame(te_ary, columns=te.columns_)

frequent_itemsets = apriori(trans_df, min_support=0.02, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.4)

rules_win = rules[rules["consequents"].astype(str).str.contains("result_win")].copy()
rules_win = rules_win.sort_values("lift", ascending=False)

def fs_to_str(fs):
    return ", ".join(sorted(list(fs)))

player_rules_df = rules_win.copy()
player_rules_df["antecedent"] = player_rules_df["antecedents"].apply(fs_to_str)
player_rules_df["consequent"] = player_rules_df["consequents"].apply(fs_to_str)

player_rules_df = player_rules_df[
    ["antecedent", "consequent", "support", "confidence", "lift"]
].reset_index(drop=True)

player_rules_df.head()
```

```python
import pandas as pd
from itertools import combinations
from mlxtend.frequent_patterns import apriori, association_rules

# ===== 1) Chuẩn hoá player_df =====
cols_needed = ["gameid", "teamname", "position", "champion", "result"]
for c in cols_needed:
    assert c in player_df.columns, f"Thiếu cột {c} trong player_df"

df = player_df[cols_needed].copy()
df["gameid"] = df["gameid"].astype(str)
df["teamname"] = df["teamname"].astype(str)
df["position"] = df["position"].astype(str).str.upper()
df["champion"] = df["champion"].astype(str)
df["result"] = df["result"].astype(int)

# ===== 2) Tách theo vị trí =====
mid_df = df[df["position"] == "MID"]
jng_df = df[df["position"].isin(["JNG", "JUNGLE"])]
bot_df = df[df["position"].isin(["ADC", "BOT", "BOTTOM"])]
sup_df = df[df["position"].isin(["SUP", "SUPPORT"])]

# ===== 3) Hàm tạo cặp trong cùng team =====
def build_pair_df(df_pos_a, df_pos_b, label):
    merged = df_pos_a.merge(
        df_pos_b,
        on=["gameid", "teamname"],
        suffixes=("_a", "_b"),
    )
    pairs = merged[["gameid", "teamname", "champion_a", "champion_b", "result_a"]].copy()
    pairs.rename(
        columns={
            "champion_a": f"{label}_1",
            "champion_b": f"{label}_2",
            "result_a": "win",
        },
        inplace=True,
    )
    return pairs

mid_jng_pairs = build_pair_df(mid_df, jng_df, "champ")
bot_sup_pairs = build_pair_df(bot_df, sup_df, "champ")

# ===== 4) Tổng hợp winrate cặp MID–JNG, BOT–SUP =====
def aggregate_pairs(pairs_df, col1, col2):
    agg = (
        pairs_df
        .groupby([col1, col2], as_index=False)
        .agg(
            games=("win", "size"),
            wins=("win", "sum"),
        )
    )
    agg["winrate"] = agg["wins"] / agg["games"]
    return agg

mid_jng_stats = aggregate_pairs(mid_jng_pairs, "champ_1", "champ_2")
bot_sup_stats = aggregate_pairs(bot_sup_pairs, "champ_1", "champ_2")

min_games = 10

mid_jng_stats = (
    mid_jng_stats[mid_jng_stats["games"] >= min_games]
    .sort_values(by=["winrate", "games"], ascending=[False, False])
    .reset_index(drop=True)
)
bot_sup_stats = (
    bot_sup_stats[bot_sup_stats["games"] >= min_games]
    .sort_values(by=["winrate", "games"], ascending=[False, False])
    .reset_index(drop=True)
)

mid_jng_stats["winrate"] = mid_jng_stats["winrate"].round(2)
bot_sup_stats["winrate"] = bot_sup_stats["winrate"].round(2)

mid_jng_stats = mid_jng_stats.rename(
    columns={"champ_1": "mid_champ", "champ_2": "jng_champ"}
)
bot_sup_stats = bot_sup_stats.rename(
    columns={"champ_1": "bot_champ", "champ_2": "sup_champ"}
)

# ===== 5) Luật kết hợp MID–JNG =====
mid_jng = mid_df.merge(
    jng_df,
    on=["gameid", "teamname"],
    suffixes=("_mid", "_jng"),
)

wins_mj = mid_jng[mid_jng["result_mid"] == 1].copy()
wins_mj["item_mid"] = "MID_" + wins_mj["champion_mid"].astype(str)
wins_mj["item_jng"] = "JNG_" + wins_mj["champion_jng"].astype(str)

transactions_mj = wins_mj[["item_mid", "item_jng"]]
all_items_mj = sorted(
    set(transactions_mj["item_mid"]).union(set(transactions_mj["item_jng"]))
)

def encode_row_mj(row):
    return [1 if item in row.values else 0 for item in all_items_mj]

encoded_mj = transactions_mj.apply(encode_row_mj, axis=1, result_type="expand")
encoded_mj.columns = all_items_mj

freq_itemsets_mj = apriori(encoded_mj, min_support=0.01, use_colnames=True)
rules_all_mj = association_rules(freq_itemsets_mj, metric="confidence", min_threshold=0.05)

def has_mid(items): return any(i.startswith("MID_") for i in items)
def has_jng(items): return any(i.startswith("JNG_") for i in items)

rules_mid_to_jng = rules_all_mj[
    rules_all_mj["antecedents"].apply(has_mid) &
    rules_all_mj["consequents"].apply(has_jng)
].copy()

def unpack_rule_mj(row):
    a = next(iter(row["antecedents"]))
    c = next(iter(row["consequents"]))
    return pd.Series(
        {
            "pair_type": "MID_JNG",
            "role1": "MID", "champ1": a.replace("MID_", ""),
            "role2": "JNG", "champ2": c.replace("JNG_", ""),
            "support": row["support"],
            "confidence": row["confidence"],
            "lift": row["lift"],
        }
    )

rules_mid_df = rules_mid_to_jng.apply(unpack_rule_mj, axis=1)

# ===== 6) Luật kết hợp BOT–SUP =====
bot_sup = bot_df.merge(
    sup_df,
    on=["gameid", "teamname"],
    suffixes=("_bot", "_sup"),
)

wins_bs = bot_sup[bot_sup["result_bot"] == 1].copy()
wins_bs["item_bot"] = "BOT_" + wins_bs["champion_bot"].astype(str)
wins_bs["item_sup"] = "SUP_" + wins_bs["champion_sup"].astype(str)

transactions_bs = wins_bs[["item_bot", "item_sup"]]
all_items_bs = sorted(
    set(transactions_bs["item_bot"]).union(set(transactions_bs["item_sup"]))
)

def encode_row_bs(row):
    return [1 if item in row.values else 0 for item in all_items_bs]

encoded_bs = transactions_bs.apply(encode_row_bs, axis=1, result_type="expand")
encoded_bs.columns = all_items_bs

freq_itemsets_bs = apriori(encoded_bs, min_support=0.01, use_colnames=True)
rules_all_bs = association_rules(freq_itemsets_bs, metric="confidence", min_threshold=0.05)

def has_bot(items): return any(i.startswith("BOT_") for i in items)
def has_sup(items): return any(i.startswith("SUP_") for i in items)

rules_bot_to_sup = rules_all_bs[
    rules_all_bs["antecedents"].apply(has_bot) &
    rules_all_bs["consequents"].apply(has_sup)
].copy()

def unpack_rule_bs(row):
    a = next(iter(row["antecedents"]))
    c = next(iter(row["consequents"]))
    return pd.Series(
        {
            "pair_type": "BOT_SUP",
            "role1": "BOT", "champ1": a.replace("BOT_", ""),
            "role2": "SUP", "champ2": c.replace("SUP_", ""),
            "support": row["support"],
            "confidence": row["confidence"],
            "lift": row["lift"],
        }
    )

rules_bot_df = rules_bot_to_sup.apply(unpack_rule_bs, axis=1)

# ===== 7) Gộp & làm tròn luật kết hợp =====
pair_rules_df = pd.concat([rules_mid_df, rules_bot_df], ignore_index=True)
pair_rules_df[["support", "confidence", "lift"]] = pair_rules_df[["support", "confidence", "lift"]].round(3)
pair_rules_df = pd.concat([rules_mid_df, rules_bot_df], ignore_index=True)
pair_rules_df[["support", "confidence", "lift"]] = pair_rules_df[["support", "confidence", "lift"]].round(3)

# Sắp theo support giảm dần, sau đó confidence, rồi lift
pair_rules_df = pair_rules_df.sort_values(
    by=["support", "confidence", "lift"],
    ascending=[False, False, False]
).reset_index(drop=True)


mid_jng_stats.head(), bot_sup_stats.head(), pair_rules_df.head()
```
